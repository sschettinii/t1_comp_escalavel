## Exercício 4

Implemente manualmente a função:

`def mean(values: Array[Double]): Double`

Use `while`, sem usar `values.sum`.

````scala
def mean(values: Array[Double]): Double = {
    var sum: Double = 0.0
    var i: Int = 0

    while (i < values.length) {
        sum += values(i)
        i += 1
    }

    sum / values.length
}
````

## Exercício 7

Implemente manualmente, usando `while`:

```scala
def maximum(values: Array[Double]): Double
```

A função deve devolver o maior elemento. Considere que o array não pode estar vazio.

Exemplo:

```scala
maximum(Array(2.0, 9.0, 4.0))
```

Resultado esperado: `9.0`.

````scala
def maximum(values: Array[Double]): Double = {
    var index: Int = 0
    var currentMaxValue: Double = values(0)
    while (index < values.length) {
        if (values(index) > currentMaxValue) currentMaxValue = values(index)
        index += 1
    }
    currentMaxValue
}

maximum(Array(2.0, 9.0, 4.0))
````

## Exercício 10


Exemplo:

```scala

```

Resultado esperado: `3`.

````scala
def countLabel(labels: Array[Int], target: Int): Int = {
    var index: Int = 0
    var count: Int = 0 
    while (index < labels.length) {
        if (labels(index) == target) count += 1
        index += 1
    }
    count
}

countLabel(Array(0, 1, 1, 0, 1), 1)
````

## Exercício 14

Implemente usando `map`:

```scala
def labelsOf(samples: Array[Sample]): Array[Int]
```

A função deve criar um novo array contendo apenas os rótulos das amostras.

Exemplo:

```scala
val samples = Array(
  Sample(Array(1.0, 2.0), 0),
  Sample(Array(3.0, 4.0), 1),
  Sample(Array(5.0, 6.0), 1)
)

labelsOf(samples)
```

Resultado esperado: `Array(0, 1, 1)`.

````scala
def labelsOf(samples: Array[Sample]): Array[Int] = {
    var labels: Array[Int] = Array(0, 0, 0)
    var index: Int = 0
    while (index < samples.length) {
        labels(index) = samples(index).label
        index += 1
    }
    labels
}

val samples = Array(
  Sample(Array(1.0, 2.0), 0),
  Sample(Array(3.0, 4.0), 1),
  Sample(Array(5.0, 6.0), 1)
)

labelsOf(samples)
````
